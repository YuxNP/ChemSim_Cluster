# Similarity Algorithms
## This method compares MS2 fragmentation patterns between related molecules to identify fine-scale structural variation.
### The full codes and example raw data will be uploaded after the article is published. The currently shown data has had the sensitive parts removed.

# Introduction:
#### Our goal was to understand the diversity of the possible peptide analogues using the LC-MS/MS information. While LC-MS is a widely used tool for analyzing complex mixtures and identifying small molecules, its potential for resolving specific structural differences between analogues remains underexplored. Existing methods, such as molecular networking or annotation-based tools like SIRIUS, are often designed for broader structural annotation or grouping related compounds, rather than identifying exact points of variation between analogues. In this study, we introduce a framework that combines LC-MS data with similarity-based algorithms to identify specific molecular changes between analogues. Large numbers of samples were analyzed via LC-HRMS, and MS fragment lists were compiled and processed. The peak lists were normalized and initially analyzed using the Bray-Curtis similarity index to exclude low-quality and non-target MS features, such as adducts. The selected analogue MS features were then analyzed using Jaccard similarity with transformed binary data to assess fragmentation pattern similarity, enabling the identification of precise points of variation in complex molecules such as vitilevuamide (example shown below)

#### A large set of samples was analyzed via LC-HRMS, and corresponding MS² fragment lists were curated. To ensure robustness, all peak lists were normalized prior to similarity analysis. Bray-Curtis dissimilarity was first applied to filter out non-informative features (e.g., adducts and low-abundance noise). High-quality analogue features were then subjected to Jaccard similarity analysis using binarized MS² spectra, enabling fine-scale comparison of fragmentation patterns across analogues.

## Example -- Vitilevuade:

### Introduction:
#### Vitilevuamide (1) is a structurally unique bicyclic tetradecapeptide elaborated with an N-terminal succinic acid, which contains Lan, Dha, two homo-isoleucine (Hil) residues,15 and N-methyl-γ-methoxythreonine (Mmt). We detected 1 with other minor analogues by LCMS in a range of didemnid tunicate specimens, including invasive species, collected in Nelson Mandela Bay (previously Algoa Bay), South Africa. Due to their low abundance, these minor analogues could not be isolated or structurally characterized using standard techniques like NMR or X-ray crystallography.

### Solution: 
#### To clarify points of structural variation between vitilevuamide analogues, it was necessary to assign key MS fragments observed for 1. The neutral losses between MS2 fragments were calculated using a difference matrix and compared with the component canonical and non-canonical amino acid residues in Vitilevuamide, as well as succinate, using a threshold error of 5 (gold), 10 (silver), and 20 (bronze) ppm to filter potential matches at different confidence levels (Table 1). The result indicated that 1 yielded distinct fragmentations with a transformed b/y ion fragmentation series attributable to the upper cycle (I) of the bicyclic system.

### Result and discussion:
#### To compare the fragmentation patterns of structurally related analogues, a two-step similarity analysis was implemented, combining Bray-Curtis and Jaccard similarity, followed by Principal Coordinates Analysis (PCoA) for visualization. Each MS/MS spectrum was normalized individually by scaling fragment ion intensities relative to the highest intensity peak within the spectrum to minimize the impact of ionization efficiency differences and ensured that comparisons were based on peak presence rather than absolute intensity. The Bray-Curtis similarity considers both the presence and relative abundance of fragment ions, making it suitable for distinguishing major structural differences. Spectra exhibiting the highest dissimilarity were removed to ensure that only structurally relevant analogues were included in further analysis. Following the initial filtering step, the Jaccard similarity coefficient was used to assess fragment presence and absence among analogues. This binary-based metric enabled precise identification of shared and unique fragment ions, facilitating the localization of structural modifications between related compounds. PCoA was performed on the Jaccard similarity matrix to visualize the structural relationships between analogues in a lower-dimensional space. This approach provided insight into clustering patterns based on fragmentation similarities and highlighted regions of structural variation.
#### At least 24 unique vitilevuamide MS features representing up to 19 analogues were discovered from classical molecular networking  of LCMS2 data for seven tunicate extracts using GNPS, an open-access knowledge base and community platform for sharing, analyzing, and annotating mass spectrometry data for small molecule metabolites21. Three molecular subnetworks of vitilevuamide derivatives were evident, comprising multiple nodes with the same m/z value due to fragmentations at three different collision energies. Subnetwork 1 comprises 23 nodes with 8 different m/z values ranging from m/z 1573.80 to m/z 1620.83; subnetwork 2 has 39 nodes with m/z values between m/z 1625.85 and 1835.91; subnetwork 3 has only three nodes attributable to the same molecule with m/z 1619.81. A few single nodes were also selected because their MS2 spectra contained either the fragment m/z 423.26 or 632.28. In order to localize the structural differences within the closely related molecules represented in the vitilevuamide subnetworks, we turned to statistical analyses of the MS2 data for select vitilevuamide analogues.



#### To ensure the highest quality data, MS1 features that yielded either the MS2 fragment m/z 423.26 or 1058.50 were curated and MS2 peaks for each feature were normalized internally. Possible analogues with poor fragmentation (i.e., too few fragmentation peaks or low-intensity peaks) were excluded. Twenty natural analogues represented in the vitilevuamide subnetworks, including 1, were selected for further analysis based on their retention time, scan number and m/z value. Since these are large molecules, fragments smaller than m/z 300 could be misleading because they derive from more than one region in the peptide ring system. All MS2 fragments greater than m/z 300 were analyzed statistically for their number of occurrences across these 19 viti-family molecules. The next step is by analyzing MS2 fragmentation intensities, grouping them based on proximity within a defined threshold, and calculating the occurrences of each fragmentation along with their average values and intensities. 

#### From the analysis, the most common and key MS2 ion was m/z 423.258, which indicated the Dha-Hil-2-Val-[Mmt-H2O] y ion for cycle II. It occurred in 16/19 analogues, whereas m/z 441.2690 for Dha-Hil-2-Val-Mmt only occurred in 6/19 analogues.  Presence/absence and relative abundance of these two ions serve as evidence that a majority of analogues differ at the Mmt residue in cycle II.
#### While most analogues exhibited a consistent series of MS² fragments representing the complete upper cycle I—typically including ions at m/z ~1058, 945, 779, and 632—a few exceptions deviated from this pattern. Notably, one feature at m/z 1619.8001 lacked the m/z 632 fragment, suggesting a structural difference in ring I relative to the parent compound. A nearly isobaric analogue at m/z 1619.8101 was initially mistaken for the same compound, but distinct fragmentation patterns and mass defects indicated they were separate molecules. The analogue at m/z 1619.8001 showed fragments consistently 2 Da lower than expected, suggesting a dehydration event likely involving a malic acid moiety. Conversely, the 1619.8101 analogue retained typical cycle I fragments but lacked the m/z 441 fragment, implying a substitution at residue 13, potentially with Phe. Additional analogues at m/z 1583.83 and 1625.85 also lacked typical ring I ions, instead showing fragment shifts of ~20 Da. These mass differences, along with the presence or absence of Mmt-related fragments, support modifications at residues 6 and 13, likely involving substitution of the Phe sidechain with a nonpolar, saturated hydrocarbon consistent with Hil.



## Future Direction
### However, we acknowledge that under certain conditions—such as when peptides and their analogues are highly repeated and replaced with similar residues like Ile and Leu and the size of the peptides are relatively small, resulting in an abundance of non-unique fragments and insufficient numbers of data points—this method may not yield reliable separation results. This limitation can be observed, for example, in fungal-encoded peptides such as paraisariamides. Many of these analogues have symmetrical structures with no unique residues, and the biosynthetic pathway also allows certain residues to be freely exchanged within multiple blocks. Consequently, some fragments share similar sequences and can occasionally yield identical fragment masses, thereby complicating the separation process or leading to incorrect justifications during grouping. 
### To address the questions above, several other similarity measures were applied and compared with the current Jaccard method. The results indicated that there was no significant difference between the Jaccard and Dice similarity measures, both of which focus on mathematical distances between similar components. Additionally, the cosine similarity measure, which compares the angular relationship between two vectors, also yielded similar results. These findings suggest that the sample preparation method used in previous analyses has a significant impact on the similarity outcomes.
